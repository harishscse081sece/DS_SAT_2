```python

from google.colab import files
import pandas as pd
import os

uploaded = files.upload()

# Load the five datasets
passengers = pd.read_csv("passengers.csv")
flights = pd.read_csv("flights.csv")
delays = pd.read_csv("delays.csv")
service_scores = pd.read_csv("service_scores.csv")
feedback = pd.read_csv("feedback.csv")

datasets = {
    "passengers": passengers,
    "flights": flights,
    "delays": delays,
    "service_scores": service_scores,
    "feedback": feedback
}

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    display(df.head())
```

```python

for name, df in datasets.items():
    print("\n", "=" * 40)
    print("DATASET:", name.upper())
    print("=" * 40)

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nNumerical summary:")
    display(df.describe(include="all").T)
```

```python

import numpy as np

# Standardize column names
for name in datasets:
    df = datasets[name].copy()
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )
    datasets[name] = df.drop_duplicates().copy()

# Update dataset variables
passengers = datasets["passengers"]
flights = datasets["flights"]
delays = datasets["delays"]
service_scores = datasets["service_scores"]
feedback = datasets["feedback"]

# Standardize class names if a class column exists
for df in datasets.values():
    if "class" in df.columns:
        df["class"] = (
            df["class"].astype("string")
            .str.strip().str.lower()
            .replace({
                "business class": "business",
                "economy class": "economy",
                "first class": "first"
            })
        )

# Identify extreme delays without deleting the original values
for col in ["departure_delay", "arrival_delay"]:
    if col in delays.columns:
        delays[col] = pd.to_numeric(delays[col], errors="coerce")
        q1 = delays[col].quantile(0.25)
        q3 = delays[col].quantile(0.75)
        iqr = q3 - q1
        delays[col + "_extreme"] = (
            (delays[col] < q1 - 1.5 * iqr) |
            (delays[col] > q3 + 1.5 * iqr)
        )

print("Data cleaning completed.")
for name, df in datasets.items():
    print(name, "shape:", df.shape)
```

```python

# Try common ID columns to combine the datasets
merged = datasets["passengers"].copy()

id_priority = [
    "passenger_flight_id",
    "passenger_id",
    "flight_id",
    "booking_id",
    "record_id"
]

for name in ["flights", "delays", "service_scores", "feedback"]:
    df = datasets[name].copy()

    # Find shared ID columns
    shared_ids = [
        col for col in id_priority
        if col in merged.columns and col in df.columns
    ]

    # Also detect other shared ID columns
    extra_ids = [
        col for col in merged.columns.intersection(df.columns)
        if col.endswith("_id") and col not in shared_ids
    ]
    shared_ids += extra_ids

    if not shared_ids:
        print(
            f"Cannot automatically merge {name}. "
            f"Check its columns and the ID keys."
        )
        print("Current columns:", merged.columns.tolist())
        print("Next dataset columns:", df.columns.tolist())
        break

    merged = merged.merge(
        df, on=shared_ids, how="left",
        suffixes=("", f"_{name}")
    )
    print(f"Merged {name} using:", shared_ids)

display(merged.head())
print("Merged dataset shape:", merged.shape)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

service_cols = [
    col for col in service_scores.columns
    if any(word in col for word in [
        "check_in", "seat", "food", "staff", "baggage"
    ])
]

for col in service_cols:
    service_scores[col] = pd.to_numeric(
        service_scores[col], errors="coerce"
    )

print("Average service ratings:")
display(service_scores[service_cols].mean().sort_values().to_frame(
    "Average Rating"
))

print("\nService rating variation:")
display(service_scores[service_cols].std().sort_values(
    ascending=False
).to_frame("Standard Deviation"))

service_scores[service_cols].mean().sort_values().plot(
    kind="bar", figsize=(9, 5), color="steelblue"
)
plt.title("Average Ratings by Service Area")
plt.xlabel("Service Area")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Find the satisfaction column
satisfaction_candidates = [
    col for col in merged.columns
    if "satisfaction" in col
]

print("Satisfaction columns:", satisfaction_candidates)

delay_cols = [
    col for col in ["departure_delay", "arrival_delay"]
    if col in merged.columns
]

if satisfaction_candidates and delay_cols:
    satisfaction_col = satisfaction_candidates[0]

    # Convert satisfaction to numeric when possible
    merged[satisfaction_col] = pd.to_numeric(
        merged[satisfaction_col], errors="coerce"
    )

    display(
        merged.groupby(
            pd.cut(merged[delay_cols[0]], bins=5),
            observed=False
        )[satisfaction_col].mean().to_frame(
            "Average Satisfaction"
        )
    )

    sns.scatterplot(
        data=merged,
        x=delay_cols[0],
        y=satisfaction_col
    )
    plt.title("Departure Delay vs Passenger Satisfaction")
    plt.xlabel("Departure Delay")
    plt.ylabel("Satisfaction")
    plt.tight_layout()
    plt.show()
else:
    print(
        "Check the merged column names. "
        "A satisfaction column and delay column are needed."
    )
```

```python

# Compare satisfaction by class and loyalty status
for col in ["class", "loyalty_status", "route"]:
    if col in merged.columns and satisfaction_candidates:
        target = satisfaction_candidates[0]

        print(f"\nAverage satisfaction by {col}:")
        display(
            merged.groupby(col, dropna=False)[target]
            .mean()
            .sort_values(ascending=False)
            .to_frame("Average Satisfaction")
        )

        if col in ["class", "loyalty_status"]:
            merged.groupby(col, dropna=False)[target].mean().plot(
                kind="bar", figsize=(7, 4)
            )
            plt.title(f"Satisfaction by {col.replace('_', ' ').title()}")
            plt.ylabel("Average Satisfaction")
            plt.tight_layout()
            plt.show()

# Routes with the lowest satisfaction
if "route" in merged.columns and satisfaction_candidates:
    display(
        merged.groupby("route")[satisfaction_candidates[0]]
        .agg(["mean", "count"])
        .rename(columns={
            "mean": "Average Satisfaction",
            "count": "Number of Records"
        })
        .sort_values("Average Satisfaction")
        .head(10)
    )
```

```python

# Create new features
if "departure_delay" in merged.columns and "arrival_delay" in merged.columns:
    merged["total_delay"] = (
        merged["departure_delay"].fillna(0) +
        merged["arrival_delay"].fillna(0)
    )

available_service_cols = [
    col for col in service_cols if col in merged.columns
]

if available_service_cols:
    merged["average_service_rating"] = (
        merged[available_service_cols].mean(axis=1)
    )

# Encode categorical variables
encoded = pd.get_dummies(
    merged, drop_first=True, dtype=int
)

print("Engineered features:")
print([
    col for col in ["total_delay", "average_service_rating"]
    if col in merged.columns
])

print("\nEncoded dataset shape:", encoded.shape)
display(encoded.head())
```

```python

from sklearn.feature_selection import SelectKBest, f_regression

target_candidates = [
    col for col in merged.columns
    if "satisfaction" in col
]

if target_candidates:
    target = target_candidates[0]
    y = pd.to_numeric(merged[target], errors="coerce")

    # Select numerical predictors only
    X = merged.select_dtypes(include=np.number).drop(
        columns=[target], errors="ignore"
    )

    valid = y.notna()
    X = X.loc[valid].replace([np.inf, -np.inf], np.nan)
    y = y.loc[valid]

    # Fill missing predictor values
    X = X.fillna(X.median()).fillna(0)

    if len(X) > 1 and y.nunique() > 1 and X.shape[1] > 0:
        k = min(5, X.shape[1])
        selector = SelectKBest(score_func=f_regression, k=k)
        selector.fit(X, y)

        selected_features = X.columns[selector.get_support()]
        print("Selected features:")
        print(selected_features.tolist())
    else:
        print("Not enough valid numeric satisfaction data for selection.")
else:
    print("No satisfaction column found. Check the merged column names.")
```

```python

output_file = "aerovista_cleaned_dataset.csv"

merged.to_csv(output_file, index=False)

print("Cleaned dataset saved as:", output_file)

from google.colab import files
files.download(output_file)
```
